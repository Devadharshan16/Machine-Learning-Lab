2. Using SVM, a company wants to classify emails as spam or not spam based on features such as word frequency
and message length.

```python
# Importing Required Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report
from sklearn import metrics
from sklearn.preprocessing import StandardScaler
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
# Generating a Synthetic Email Spam Dataset
np.random.seed(42)
n_samples = 500

# Features: word_frequency, message_length, special_char_count, link_count
# Spam emails tend to have higher word frequency, longer messages, more special chars and links
spam_word_freq = np.random.normal(70, 15, n_samples // 2)
spam_msg_len = np.random.normal(300, 80, n_samples // 2)
spam_special_chars = np.random.normal(15, 5, n_samples // 2)
spam_link_count = np.random.normal(5, 2, n_samples // 2)

ham_word_freq = np.random.normal(35, 12, n_samples // 2)
ham_msg_len = np.random.normal(150, 60, n_samples // 2)
ham_special_chars = np.random.normal(5, 3, n_samples // 2)
ham_link_count = np.random.normal(1, 1, n_samples // 2)

word_frequency = np.concatenate([spam_word_freq, ham_word_freq])
message_length = np.concatenate([spam_msg_len, ham_msg_len])
special_char_count = np.concatenate([spam_special_chars, ham_special_chars])
link_count = np.concatenate([spam_link_count, ham_link_count])
label = np.concatenate([np.ones(n_samples // 2), np.zeros(n_samples // 2)])  # 1 = Spam, 0 = Not Spam

data = pd.DataFrame({
    'word_frequency': word_frequency,
    'message_length': message_length,
    'special_char_count': special_char_count,
    'link_count': link_count,
    'label': label.astype(int)
})
```

```python
# Display the first few rows
print(data.shape)
data.head()
```

```python
# Checking for any null values in dataset
data.isnull().sum()
```

```python
# Assigning dependent and independent variables
feature_cols = ['word_frequency', 'message_length', 'special_char_count', 'link_count']
x = data[feature_cols]
y = data.label
```

```python
# Splitting the dataset into Training and Testing Dataset
x_train, x_test, y_train, y_test = train_test_split(x, y,
      test_size=0.3, random_state=5)

display(x_train.shape, y_train.shape, x_test.shape,
      y_test.shape)
```

```python
# Preprocessing Data with StandardScaler
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
# Fitting the Model (SVM) using 'rbf' kernel
model = SVC(kernel='rbf', random_state=0)
model.fit(x_train, y_train)
svc_prediction = model.predict(x_test)
print('svc_prediction: ', svc_prediction)
```

```python
# Evaluation Metrics for 'rbf' kernel
conf_mat = metrics.confusion_matrix(y_test, svc_prediction)
print('SVC [ kernel - rbf ]')
print('Confusion Matrix : \n', conf_mat)
Accuracy_score = metrics.accuracy_score(y_test, svc_prediction)
print('Accuracy Score : ', Accuracy_score)
print('Accuracy in Percentage : ', int(Accuracy_score * 100), '%')
print(classification_report(svc_prediction, y_test))
```

```python
# Confusion Matrix Heatmap for 'rbf' kernel
conf_mat = pd.crosstab(y_test, svc_prediction, rownames=['Actual'],
      colnames=['Predicted'])
sn.heatmap(conf_mat, annot=True).set_title('SVC [rbf]')
```

```python
# Fitting the Model (SVM) using 'linear' kernel
model = SVC(kernel='linear', random_state=0)
model.fit(x_train, y_train)
svc_prediction = model.predict(x_test)
print('svc_prediction: ', svc_prediction)
```

```python
# Evaluation Metrics for 'linear' kernel
conf_mat = metrics.confusion_matrix(y_test, svc_prediction)
print('SVC [ kernel - linear ]')
print('Confusion Matrix : \n', conf_mat)
Accuracy_score = metrics.accuracy_score(y_test, svc_prediction)
print('Accuracy Score : ', Accuracy_score)
print('Accuracy in Percentage : ', int(Accuracy_score * 100), '%')
print(classification_report(svc_prediction, y_test))
```

```python
# Confusion Matrix Heatmap for 'linear' kernel
conf_mat = pd.crosstab(y_test, svc_prediction, rownames=['Actual'],
      colnames=['Predicted'])
sn.heatmap(conf_mat, annot=True).set_title('SVC [linear]')
```
