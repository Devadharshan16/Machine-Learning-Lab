2. Calculate the distance between a new entry and other existing values using the Euclidean distance formula.

| Brightness | Saturation | Class |
|------------|------------|-------|
| 40 | 20 | Red |
| 50 | 50 | Blue |
| 60 | 90 | Blue |
| 10 | 25 | Red |
| 70 | 70 | Blue |
| 60 | 10 | Red |
| 25 | 80 | Blue |

```python
# Importing Required Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report
from sklearn import metrics
from sklearn.preprocessing import StandardScaler
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
# Creating the Dataset from the given table
data = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
})

print(data.shape)
data.head()
```

```python
# Checking for any null values in dataset
data.isnull().sum()
```

```python
# Encoding the Class labels
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data['Class_encoded'] = le.fit_transform(data['Class'])
print('Encoded Classes:', dict(zip(le.classes_, le.transform(le.classes_))))
data
```

```python
# Assigning dependent and independent variables
x = data[['Brightness', 'Saturation']]
y = data['Class_encoded']
```

```python
# Defining a new entry to classify
new_entry = np.array([[45, 55]])
print('New Entry - Brightness: 45, Saturation: 55')
```

```python
# Calculating Euclidean Distance between the new entry and all existing values
distances = []
for i in range(len(data)):
    dist = np.sqrt((data['Brightness'].iloc[i] - new_entry[0][0])**2 +
                   (data['Saturation'].iloc[i] - new_entry[0][1])**2)
    distances.append(dist)

data['Distance_from_new'] = distances
print('Euclidean Distances from new entry (45, 55):')
data[['Brightness', 'Saturation', 'Class', 'Distance_from_new']]
```

```python
# Sorting by distance to find nearest neighbors
sorted_data = data.sort_values('Distance_from_new')
print('Sorted by Distance (Nearest Neighbors):')
sorted_data[['Brightness', 'Saturation', 'Class', 'Distance_from_new']]
```

```python
# Training the Model using KNN (minkowski)
results = []
for i in [1, 2, 3, 4, 5]:
    model = KNeighborsClassifier(n_neighbors=i, metric='minkowski', p=2)
    model.fit(x, y)
    y_pred = model.predict(x)
    Accuracy_score = metrics.accuracy_score(y, y_pred)
    results.append(Accuracy_score)

print('KNN [ minkowski ]')
print('for n_neighbor=3 : ')
model_k3 = KNeighborsClassifier(n_neighbors=3, metric='minkowski', p=2)
model_k3.fit(x, y)
y_pred = model_k3.predict(x)
conf_mat = metrics.confusion_matrix(y, y_pred)
print('\n Confusion Matrix : ', conf_mat)
Accuracy_score = metrics.accuracy_score(y, y_pred)
print('Accuracy Score : ', Accuracy_score)
print('Accuracy in Percentage : ', int(Accuracy_score*100), '%')
print('\n', classification_report(y_pred, y))

print(results)
```

```python
# Predicting the class of the new entry
prediction = model_k3.predict(new_entry)
predicted_class = le.inverse_transform(prediction)
print(f'Predicted class for new entry (Brightness=45, Saturation=55): {predicted_class[0]}')
```

```python
# Visualization - Scatter plot with new entry
plt.figure(figsize=(8, 6))
colors = {'Red': 'red', 'Blue': 'blue'}
for cls in data['Class'].unique():
    subset = data[data['Class'] == cls]
    plt.scatter(subset['Brightness'], subset['Saturation'],
                c=colors[cls], label=cls, s=100, edgecolors='black')

plt.scatter(new_entry[0][0], new_entry[0][1], c='green', marker='*',
            s=300, label=f'New Entry ({predicted_class[0]})', edgecolors='black')
plt.xlabel('Brightness')
plt.ylabel('Saturation')
plt.title('KNN Classification - Brightness vs Saturation')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Comparing accuracy for different values of K
models = pd.DataFrame({
    'n_neighbors': ['1', '2', '3', '4', '5'],
    'Accuracy Score': [results[0], results[1], results[2], results[3], results[4]]
})
models.sort_values(by='Accuracy Score')
print(models.to_string(index=False))
```
